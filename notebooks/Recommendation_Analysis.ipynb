```python
import pandas as pd
```

```python
items = pd.read_csv("../data/items.csv")
users = pd.read_csv("../data/users.csv")
```

```python
print("Items Dataset Shape:", items.shape)
print("Users Dataset Shape:", users.shape)
```

```python
print("Items Columns:")
print(items.columns.tolist())

print("\nUsers Columns:")
print(users.columns.tolist())
```

```python
print("Items Missing Values:")
print(items.isnull().sum())

print("\nUsers Missing Values:")
print(users.isnull().sum())
```

```python
print("Duplicate item rows:", items.duplicated().sum())
print("Duplicate user rows:", users.duplicated().sum())
```

```python
print("Duplicate Item IDs:", items["Item_ID"].duplicated().sum())
print("Duplicate User IDs:", users["User_ID"].duplicated().sum())
```

```python
print("Items Data Types:")
print(items.dtypes)

print("\nUsers Data Types:")
print(users.dtypes)
```

```python
invalid_ratings = items[
    (items["Rating"] < 5.0) |
    (items["Rating"] > 9.5)
]

print("Invalid Ratings:")
print(invalid_ratings)
```

```python
invalid_categories = items[items["Category"] != "Movie"]

print("Invalid Item Categories:")
print(invalid_categories)
```

```python
invalid_user_categories = users[
    users["Preferred_Category"] != "Movie"
]

print("Invalid User Categories:")
print(invalid_user_categories)
```

```python
print("===== DATASET VALIDATION SUMMARY =====")

print("Items shape:", items.shape)
print("Users shape:", users.shape)

print("Missing item values:", items.isnull().sum().sum())
print("Missing user values:", users.isnull().sum().sum())

print("Duplicate item rows:", items.duplicated().sum())
print("Duplicate user rows:", users.duplicated().sum())

print("Duplicate Item IDs:", items["Item_ID"].duplicated().sum())
print("Duplicate User IDs:", users["User_ID"].duplicated().sum())

print(
    "Invalid ratings:",
    ((items["Rating"] < 5.0) | (items["Rating"] > 9.5)).sum()
)

print(
    "Invalid item categories:",
    (items["Category"] != "Movie").sum()
)

print(
    "Invalid user categories:",
    (users["Preferred_Category"] != "Movie").sum()
)
```
